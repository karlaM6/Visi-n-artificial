```python
import cv2
from deepface import DeepFace

# ----------------------------------------------------------------------
# 1. Cargar el clasificador Haar Cascade para detección de rostros
# ----------------------------------------------------------------------
faceCascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)

# ----------------------------------------------------------------------
# 2. Traducción de emociones al español
# ----------------------------------------------------------------------
EMOTION_ES = {
    "angry": "Enojo",
    "disgust": "Disgusto",
    "fear": "Miedo",
    "happy": "Felicidad",
    "sad": "Tristeza",
    "surprise": "Sorpresa",
    "neutral": "Neutral",
}

# ----------------------------------------------------------------------
# 3. Iniciar captura de video desde la webcam
# ----------------------------------------------------------------------
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    raise IOError("No se pudo acceder a la cámara. Verifica que esté conectada y libre.")

font = cv2.FONT_HERSHEY_SIMPLEX

while True:
    ret, frame = cap.read()
    if not ret:
        print("No se pudo leer el frame de la cámara.")
        break

    # Convertir a escala de grises para la detección con Haar Cascade
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Detectar rostros en el frame
    faces = faceCascade.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(60, 60),
    )

    for (x, y, w, h) in faces:
        # Recortar la región del rostro para analizarla con DeepFace
        face_roi = frame[y:y + h, x:x + w]

        try:
            # Analizar la emoción dominante del rostro recortado
            result = DeepFace.analyze(
                face_roi,
                actions=["emotion"],
                detector_backend="skip",   # ya tenemos el rostro recortado, no vuelve a detectarlo
                enforce_detection=False,
            )
            # DeepFace.analyze puede devolver una lista o un dict según la versión
            prediction = result[0] if isinstance(result, list) else result
            emotion = prediction["dominant_emotion"]
            emotion_label = EMOTION_ES.get(emotion, emotion)
        except Exception as e:
            emotion_label = "No detectado"

        # Dibujar el rectángulo alrededor del rostro
        cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

        # Escribir la palabra descriptiva de la emoción encima del rectángulo
        cv2.putText(
            frame,
            emotion_label,
            (x, y - 10 if y - 10 > 10 else y + h + 25),
            font,
            0.9,
            (0, 255, 0),
            2,
            cv2.LINE_AA,
        )

    cv2.imshow("Real-Time Face Emotion Recognition", frame)

    # Salir con la tecla 'q'
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break


cap.release()

```
